```python

genre_map = {'Nudity': 'Adult', 'Sexual Content': 'Adult', 'Violent': 'Adult', 'Gore': 'Adult','Utilities': 'Software', 'Design & Illustration': 'Software', 'Animation & Modeling': 'Software',
    'Video Production': 'Software', 'Game Development': 'Software', 'Audio Production': 'Software', 'Software Training': 'Software', 'Web Publishing': 'Software', 'Photo Editing': 'Software',
    'Accounting': 'Software', 'Movie': 'Media', 'Documentary': 'Media', 'Episodic': 'Media','Short': 'Media', 'Tutorial': 'Media', '360 Video': 'Media', 'Education': 'Education'}

game_genre = ['Indie', 'Action', 'Casual', 'Adventure', 'Simulation','Strategy', 'RPG', 'Early Access', 'Free to Play','Sports', 'Racing', 'Massively Multiplayer']


def map_gen(genre_str):
    if pd.isna(genre_str):
        return genre_str
    genres = [g.strip() for g in genre_str.split(',')]
    mapped = []
    for g in genres:
        if g in game_genre:
            mapped.append(g)
        elif g in genre_map:
            mapped.append(genre_map[g])
        else:
            mapped.append(g)
    return ','.join(sorted(set(mapped)))

game_df['Genres'] = game_df['Genres'].apply(map_gen)
```

```python
emo_labels = ["anger", "disgust", "fear", "joy", "neutral", "sadness", "surprise"]
appid = []
emotion_scores = {label: [] for label in emo_labels}

def max_emo(predictions):
    per_scores = {label: [] for label in emo_labels}
    for prediction in predictions:
        sorted_pred = sorted(prediction, key=lambda x: x['label'])
        for index, label in enumerate(emo_labels):
            per_scores[label].append(sorted_pred[index]['score'])
    return {label: max(scores) for label, scores in per_scores.items()}

for i in range(10):
    appid.append(game_df.iloc[i]['AppID'])
    sentences = game_df.iloc[i]['About the game'].split(".")
    predictions = classifier(sentences)
    max_scores = max_emo(predictions)
    for label in emotion_labels:
        emotion_scores[label].append(max_scores[label])
```

```python
def truncate_text(text):
    tokens = tokenizer.tokenize(text)
    truncated = tokens[:MAX_TOKENS]
    return tokenizer.convert_tokens_to_string(truncated)
```

### **2.2. Game genre Distribution past years 2014 - 2023**

* first the dataset is filtered to release yaers from last 10 years using sorted(df['Release_year'].dropna().unique())[-10:] method.

* The genre column was split using rhw str.split(' , ') method and flatten the list.

* next the counter is used to count the frequency of each genre for the specific year and then top 10 most common genres are selected using most_common(10) method.

* finally for each year a pie chat is genarated using the plt.pie().

```python
recent_years = sorted(df['Release_year'].unique())[-10:]
for year in recent_years:
    year_data = df[df['Release_year'] == year]
    gen_split = year_data['Genres'].str.split(",")
    all_genres = []
    for genre_list in gen_split:
        all_genres.extend([g.strip() for g in genre_list])
    genre_counts = Counter(all_genres)
    top_genres = genre_counts.most_common(10)
    labels = [g[0] for g in top_genres]
    sizes = [g[1] for g in top_genres]

    plt.figure(figsize=(8, 8))
    plt.pie(sizes,labels=labels,autopct='%1.1f%%',startangle=140,colors=sns.color_palette("pastel", len(labels)))
    plt.title(f"Genre Distribution for Games Released in the year - {year}")
    plt.show()
```

* These pie chart shows the genre distribution for games released in each years from 2014 - 2023.

* According to pie chart the highest percentange of games were indie game which was between 30 - 25%. Same as the prevoius genre plots the top 5 genres have remained the same over the 10 year period.

* By identify these patterns insightcorp can target marketing and inventory decisions based on past trends.

### **Average Emotion scores across all games**

* First a list of emotions was created and stored in the variable emo_labels.

* Next the average score for each emotion was calculated using the mean() method and sorted in descending order.
* A barplot was then created using sns.barplot(), where the x axis represents the emotion labels and the y axis represents the corresponding average scores.

```python
emo_labels = ["anger", "disgust", "fear", "joy", "neutral", "sadness", "surprise"]
emo_avg = df[emo_labels].mean().sort_values(ascending=False)
plt.figure(figsize=(10,6))
sns.barplot(x=emo_avg.index, y=emo_avg.values, palette='rocket')
plt.title('Average Emotion Scores Across All Games')
plt.ylabel('Average Score')
plt.xlabel('Emotion')
plt.show()
```

### **Average Emotion socres by year from 2014 - 2023**

* First, the years were sorted using the sorted() function on the Release_year column.

* And then a for loop was  used to generate barplots for each year using sns.barplot().

```python
recent_years = sorted(df['Release_year'].unique())[-10:]
emo_labels = ["anger", "disgust", "fear", "joy", "neutral", "sadness", "surprise"]

for year in recent_years:
    year_data = df[df['Release_year'] == year]
    avg_scores = year_data[emo_labels].mean()
    plt.figure(figsize=(10, 6))
    sns.barplot(x=emo_labels, y=avg_scores.values, palette="viridis")
    plt.title(f"Average Emotion Scores in {year}")
    plt.xlabel("Emotion")
    plt.ylabel("Average Score")
    plt.ylim(0, 1)
    plt.show()
```
